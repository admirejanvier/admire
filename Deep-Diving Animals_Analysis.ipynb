```python
divers = pd.read_csv('deepest-diving-animals.csv')

divers
```

Count animals in each category

```python
divers['category'].value_counts()
```

Find the maximum diving depth per category

```python
max_depths = divers.groupby('category')['depth'].max()

max_depths
```

Convert the Series into a DataFrame.

```python
df = max_depths.reset_index(name='max_depth')

df
```

the  default bar chart

```python
plt.bar(df['category'], df['max_depth'])

plt.ylabel('Maximum Depth (meters)')

plt.show()
```

o Create a horizontal bar chart.



o Label the x axis.


```python
plt.barh(df['category'], df['max_depth'])
plt.xlabel('Maximum Depth (meters)')
```

• Sort max_depth values.


• Display the result

```python
df = df.sort_values('max_depth')
df #display it
```

redraw the barh() and we'll get sorted bars

```python
plt.barh(df['category'], df['max_depth'])
plt.xlabel('Maximum Depth (meters)')
```

Clean the horizontal bar chart

Remove all spines.



Add faded vertical grid lines.



Hide all ticks.

```python
def clean_bar_axes():
    ax = plt.gca()

    ax.spines[['top', 'bottom', 'left', 'right']].set_visible(False)

    ax.grid(axis='x', color='black', alpha=0.5)

    ax.tick_params(axis='both', length=0)
```

**now call it to see the result**

```python
plt.barh(df['category'], df['max_depth'])

plt.xlabel('Maximum Depth (meters)')

clean_bar_axes()

plt.show()
```

**Add a color column to our DataFrame.**

Add a color column to df.



Display the result.

```python
df['color'] = 'C0'

df
```

**Add a reference row for the submarine implosion.**

o Add a reference row with a ref_0 index.

o Redo the numerical sorting.

o Display the result

```python
df.loc['ref_0'] = ['submarine implosion', 730, 'C1']

df = df.sort_values('max_depth')

df
```

**Plotting with color**

o Draw barh with the color argument.

o Label the x axis.

o Enhance the graph using clean_bar_axes()

```python
plt.barh(df['category'], df['max_depth'], color=df['color'])

plt.xlabel('Maximum Diving Depth (meters)')

clean_bar_axes()

plt.show()
```
